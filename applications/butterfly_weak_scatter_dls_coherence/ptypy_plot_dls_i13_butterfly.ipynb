## Weakly Scattering Object (DLS, I13, butterfly)

```python
import h5py
import time
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as colors
import ptypy.utils as u
```

# Loading reconstructions from .ptyr (HDF5) files

```python
path_to_recons_sdr = "/dls/science/users/iat69393/ptypy-paper/output/i13_butterfly_SDR_cupy_0500.ptyr"
path_to_recons_epie = "/dls/science/users/iat69393/ptypy-paper/output/i13_butterfly_EPIE_cupy_0500.ptyr"
path_to_recons_ml = "/dls/science/users/iat69393/ptypy-paper/output/i13_butterfly_ML_cupy_0500.ptyr"
```

```python
with h5py.File(path_to_recons_sdr, "r") as f:
    obj1 = u.rmphaseramp(f["content/obj/Sscan_00G00/data"][0,198:-398,350:-389])
    prb1 = f["content/probe/Sscan_00G00/data"][:]
    Niter1 = len(f['content/runtime/iter_info/'])
    errors1 = np.array([f["content/runtime/iter_info/%05d/error" %i][:] for i in range(Niter1)])
    niters1 = np.array([f["content/runtime/iter_info/%05d/iteration" %i][...] for i in range(Niter1)])
    pixelsize = f['content/obj/Sscan_00G00/_psize'][0]*1e9
with h5py.File(path_to_recons_epie, "r") as f:
    obj2 = u.rmphaseramp(f["content/obj/Sscan_00G00/data"][0,198:-398,350:-389])
    prb2 = f["content/probe/Sscan_00G00/data"][:]
    Niter2 = len(f['content/runtime/iter_info/'])
    errors2 = np.array([f["content/runtime/iter_info/%05d/error" %i][:] for i in range(Niter2)])
    niters2 = np.array([f["content/runtime/iter_info/%05d/iteration" %i][...] for i in range(Niter2)])
with h5py.File(path_to_recons_ml, "r") as f:
    obj3 = u.rmphaseramp(f["content/obj/Sscan_00G00/data"][0,198:-398,350:-389])
    prb3 = f["content/probe/Sscan_00G00/data"][:]
    Niter3 = len(f['content/runtime/iter_info/'])
    errors3 = np.array([f["content/runtime/iter_info/%05d/error" %i][:] for i in range(Niter3)])
    niters3 = np.array([f["content/runtime/iter_info/%05d/iteration" %i][...] for i in range(Niter3)])
e,v = u.ortho(prb3)
```

```python
offset = +0.02
obj1 *= np.exp(-1j*np.median(np.angle(obj1[:200,:200]))+1j*offset)
obj2 *= np.exp(-1j*np.median(np.angle(obj2[:200,:200]))+1j*offset)
obj3 *= np.exp(-1j*np.median(np.angle(obj3[:200,:200]))+1j*offset)
```

## Figure comparing results from different engines

```python
fig, axd = plt.subplot_mosaic([["A", "A", "B", "B", "C", "C"],
                               ["A", "A", "B", "B", "C", "C"],
                               ["A1","A2","B1","B2","C1","C2"]],
                               figsize=(12,4.5), constrained_layout=False, dpi=300,
                               gridspec_kw=dict(width_ratios=[1,1,1,1,1,1], height_ratios=[1,1,0.95]))
vmin,vmax =-0.15,0.15
cm = "magma"
im = axd["A"].imshow(-np.angle(obj2), cmap=cm, interpolation="none", vmin=vmin,vmax=vmax)
axd["A"].axis("off")
axd["A"].add_patch(plt.Rectangle((100,370),600,400, fill=0, lw=1, color="b"))
axd["A"].add_patch(plt.Rectangle((750,100),300,200, fill=0, lw=1, color="c"))
axd["A"].text(50,50, 'a', va='center', ha='center', color='w', fontsize=10, fontweight='bold')
axd["A"].add_patch(plt.Rectangle((1350,50),1e4/pixelsize,10, color='w'))
axd["A"].text(1350+1e4/pixelsize/2,80,r"%d $\mathrm{\mu}$m" %(10), color='w', va='top', ha='center', fontsize=8)

axd["B"].imshow(-np.angle(obj1), cmap=cm, interpolation="none", vmin=vmin,vmax=vmax)
axd["B"].axis("off")
axd["B"].add_patch(plt.Rectangle((100,370),600,400, fill=0, lw=1, color="b",ls=":"))
axd["B"].add_patch(plt.Rectangle((750,100),300,200, fill=0, lw=1, color="c",ls=":"))
axd["B"].text(60,60, 'b', va='center', ha='center', color='w', fontsize=10, fontweight='bold')
# axd["B"].add_patch(plt.Rectangle((1550,50),5000/pixelsize,10, color='w'))
# axd["B"].text(1550+5000/pixelsize/2,80,r"%d $\mathrm{\mu}$m" %(5000/1e3), color='w', va='top', ha='center', fontsize=10)

axd["C"].imshow(-np.angle(obj3), cmap=cm, interpolation="none", vmin=vmin,vmax=vmax)
axd["C"].axis("off")
axd["C"].add_patch(plt.Rectangle((100,370),600,400, fill=0, lw=1, color="b",ls="-."))
axd["C"].add_patch(plt.Rectangle((750,100),300,200, fill=0, lw=1, color="c",ls="-."))
axd["C"].text(50,50, 'c', va='center', ha='center', color='w', fontsize=10, fontweight='bold')
# axd["C"].add_patch(plt.Rectangle((1550,50),5000/pixelsize,10, color='w'))
# axd["C"].text(1550+5000/pixelsize/2,80,r"%d $\mathrm{\mu}$m" %(5000/1e3), color='w', va='top', ha='center', fontsize=10)

axd["A1"].imshow(-np.angle(obj2[370:770,100:700]),cmap=cm, interpolation="none",vmin=vmin,vmax=vmax)
axd["A2"].imshow(-np.angle(obj2[100:300,750:1050]),cmap=cm, interpolation="none",vmin=vmin,vmax=vmax)
axd["B1"].imshow(-np.angle(obj1[370:770,100:700]),cmap=cm, interpolation="none",vmin=vmin,vmax=vmax)
axd["B2"].imshow(-np.angle(obj1[100:300,750:1050]),cmap=cm, interpolation="none",vmin=vmin,vmax=vmax)
axd["C1"].imshow(-np.angle(obj3[370:770,100:700]),cmap=cm, interpolation="none",vmin=vmin,vmax=vmax)
axd["C2"].imshow(-np.angle(obj3[100:300,750:1050]),cmap=cm, interpolation="none",vmin=vmin,vmax=vmax)

for a,c,s in zip(["A1","A2","B1","B2","C1","C2"],["b","c","b","c","b","c"],["-","-",":",":","-.","-."]):
    axd[a].set_xticks([])
    axd[a].set_yticks([])
    for x in ["bottom","top","right","left"]:
        axd[a].spines[x].set_color(c)
        axd[a].spines[x].set_linestyle(s)
        axd[a].spines[x].set_linewidth(2)

cax = fig.add_axes([0.91,0.39,0.01,0.485])
cb = fig.colorbar(im, cax=cax)
cax.set_ylabel("Phase shift (rad)")
cax.set_yticks([vmin,0,vmax])
cb.outline.set_linewidth(0)
cax.tick_params(labelsize=9, direction="in", length=9, color="w")
plt.savefig("/dls/science/users/iat69393/ptypy-paper/figures/fig_dls_i13_butterfly.png", dpi=300, bbox_inches="tight")
plt.show()
```

## Performance metrics

```python
import json
nframes = 1260
npixel = 512
npixel2 = npixel**2
niterations = 500
```

### ML

```python
with open("./summary_dls_i13_butterfly_ml.json", "r") as f:
    metrics = json.load(f)

host = metrics["host"]
total = sum([v for v in metrics["benchmark"].values()])
time_prep_per_pixel = metrics["benchmark"]["data_load"] / nframes / npixel2  * 1e9
time_iterate_per_pixel = metrics["benchmark"]["engine_iterate"] / nframes / npixel2 / niterations * 1e9
print(f"Total: {total:.0f} s | Preparation: {time_prep_per_pixel:.2f} ns/px | Iterate: {time_iterate_per_pixel:.2f} ns/px")
print(f"Cluster node: {host}")
```
